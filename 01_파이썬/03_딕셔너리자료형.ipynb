```python
ken = {"name" : "이광민", "gender" : "male", "hobby": "listening"}
print(ken["name"])
print(ken["gender"])
print(ken["hobby"])
```

```python
#key로 value값 얻기 - pop
ken = {'name': '이광민', 'gender': 'male', 'hobby': 'listening'}

result = ken.pop("hobby")
result
```
