```python
# 매개변수를 지정하여 호출하기

def sub(a,b):
    return a - b

result = sub(a=10, b=2)
print(result)
```

```python
# 입력값이 몇개가 될지 모를때 어떻게 함수를 만들지?

def add_many(*args):
    result = 0
    for i in args:
        result = result + i
    return result

hap = add_many(1, 2, 3, 4, 5)
print(hap)
```

```python
def add_mul(choice, *args):
    if choice == "add":
        result = 0
        for i in args:
            result = result + i
    elif choice == "mul":
        result = 1
        for i in args:
            result = result * i
    
    return result

c = add_mul("add", 10, 30, 40)
print(c)

d= add_mul("mul", 5, 3, 2)
print(d)
```

```python
# 함수의 값은 언제나 하나이다

def add_and_mul(a,b):
    return a+b, a*b

result = add_and_mul(10,20)
print(result)
```
